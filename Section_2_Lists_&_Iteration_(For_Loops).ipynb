### Step-by-Step Code Breakdown: Lists & Iteration

#### 1. Defining Lists and Inspection (`# 2.1`)
- **`revenues = [50, 60, 55]`**: A list is created using square brackets `[...]`. Lists are ordered, mutable (changeable) sequences.
- **`len(revenues)`**: Returns the total number of items in the list (`3`).
- **Zero-Based Indexing**: Python indexes items starting at `0` (e.g., `revenues[0]` is `50`, `revenues[1]` is `60`).

#### 2. Direct Iteration (`# 2.2`)
- **`for rev in revenues:`**: Loops directly through each element in the list. On each pass, the variable `rev` automatically takes the value of the current item (`50`, then `60`, then `55`).
- **Use Case**: Best when you only need to read list values and do not need their positional index.

#### 3. Index-Based Loop (`# 2.3`)
- **`range(len(revenues))`**: `len(revenues)` evaluates to `3`, so `range(3)` produces the sequence `0, 1, 2`.
- **Parallel List Access**: Using index `i` allows simultaneous access to corresponding elements from multiple lists (`revenues[i]` and `expenses[i]`).

#### 4. Formatting Float Outputs (`# 2.4`)
- **Format Specifier `:.2f`**: Used inside f-strings (`{margin:.2f}`) to round floating-point numbers to exactly **2 decimal places** for cleaner visual output.

#### 5. Dynamic List Creation (`# 2.5`)
- **`margins = []`**: Initializes an empty list.
- **`.append(value)`**: Adds a new item to the end of the list on each iteration pass. This pattern is fundamental for dynamically processing and transforming data in Python.

```python
# 2.1 Defining lists
revenues = [50, 60, 55]
expenses = [10, 30, 35]

print(f"Type: {type(revenues)}")
print(f"Length: {len(revenues)}")

# 2.2 Direct loop
print("\n--- Direct Iteration ---")
for rev in revenues:
    print(rev)

# 2.3 Index-based loop
print("\n--- Index-based Loop ---")
for i in range(len(revenues)):
    profit = revenues[i] - expenses[i]
    margin = profit * 100 / revenues[i]
    print(margin)

# 2.4 Formatted float output
print("\n--- Formatted Margins ---")
for i in range(len(revenues)):
    profit = revenues[i] - expenses[i]
    margin = profit * 100 / revenues[i]
    print(f" Margin {i}: {margin:.2f}%")

# 2.5 Dynamic list creation using .append()
margins = []
for i in range(len(revenues)):
    profit = revenues[i] - expenses[i]
    margin = profit * 100 / revenues[i]
    margins.append(margin)

print(f"\nCalculated Margins List: {margins}")
```
