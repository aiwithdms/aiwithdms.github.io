1.Variable Declaration (# 1.1)
revenue = 50 & expenses = 10: Creates two variables holding whole numbers (int). Python automatically detects whole numbers and assigns them the integer data type without needing explicit declarations (dynamic typing).
# PEP 8 Standard: Variable names use snake_case (lowercase letters with underscores) and single spaces around assignment operators (=).

```python
# 1.1 Variable declaration
revenue = 50
expenses = 10
```

2. Profit and Margin Calculations (# 1.2)
profit = revenue - expenses: Subtracting two integers produces an integer (40).
margin = profit * 100 / revenue: Performs multiplication and division.
Implicit Float Conversion: Standard division (/) in Python always produces a decimal (float), even if the mathematical result is a whole number (e.g., 80.0).
Formatted Strings (f-strings): The f"..." syntax allows embedding Python expressions inside {} curly braces for dynamic evaluation

```python
# 1.2 Profit and Margin calculations
profit = revenue - expenses
print(f"Profit: {profit}")

margin = profit * 100 / revenue
print(f"Margin: {margin}")
```

3. Inspecting Types and Memory Locations (# 1.3)
type(obj): Built-in function that returns the data type of an object (<class 'int'>, <class 'float'>, etc.).
id(obj): Returns the unique memory address (integer location in RAM) where the object is stored during execution

```python
# 1.3 Inspecting variable types and memory locations
print(f"Type of margin: {type(margin)}")
print(f"Type of revenue: {type(revenue)}")
print(f"Memory address of revenue: {id(revenue)}")
```

4. String Declarations (# 1.4)
Quote Flexibility: Strings (str) can be enclosed in single ('...') or double ("...") quotes.
Triple Quotes ('''...''' or """..."""): Used for multi-line strings. They automatically retain line breaks and indentation.
Variable Reassignment: Assigning a new value to an existing variable name overwrites its previous contents.

```python
# 1.4 Multi-line and standard strings
expense_description = "food expense"
expense_description = 'food expense'
expense_description = '''
expense for food when I attended Dubai Conference
'''
print(f"Type of description: {type(expense_description)}")
```
