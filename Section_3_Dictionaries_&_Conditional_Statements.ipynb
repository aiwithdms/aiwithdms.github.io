1. Single Dictionary Lookup (# 3.1)
Key-Value Pairs ({"key": value}): Dictionaries store structured data associated with labeled keys.
Key Access (fin["revenue"]): Values are retrieved by referencing their exact key name inside square brackets [...].

```python
# 3.1 Single dictionary lookup
fin = {"revenue": 50, "expenses": 10}
profit = fin["revenue"] - fin["expenses"]
margin = profit * 100 / fin["revenue"]
print(f"Single Dict Margin: {margin}%")
```

2. Processing a List of Dictionaries (# 3.2)
Collection of Records: Storing dictionaries inside a list ([{}, {}, {}]) represents tabular data or database records.
In-Place Dictionary Mutation: fin['margin'] = ... dynamically creates a new key ('margin') and attaches the computed value directly to each dictionary as the loop runs.

```python
# 3.2 List of dictionaries
financials_list = [
    {"revenue": 50, "expenses": 10},
    {"revenue": 60, "expenses": 30},
    {"revenue": 55, "expenses": 35}
]

for fin in financials_list:
    profit = fin["revenue"] - fin["expenses"]
    fin["margin"] = profit * 100 / fin["revenue"]

print("\nUpdated List of Dicts:")
print(financials_list)
```

3. Nested Dictionaries & .items() (# 3.3)
Nested Structure: A dictionary can contain other dictionaries as values (e.g., mapping "Q1" to {"revenue": 50, "expense": 10}).
.items() Method: Returns key-value pairs simultaneously. In for quarter, data in financials_dict.items():, quarter receives the key string (e.g., "Q1"), and data receives the inner dictionary.

```python
# 3.3 Nested dictionary with conditional check
financials_dict = {
    "Q1": {"revenue": 50, "expense": 10},
    "Q2": {"revenue": 60, "expense": 30},
    "Q3": {"revenue": 55, "expense": 35},
    "Q4": {"revenue": -10, "expense": 35},
}

print("\n--- Quarterly Financial Analysis ---")
for quarter, data in financials_dict.items():
    if data["revenue"] < 0:
        print(f"Invalid revenue for {quarter}. Skipping.")
        continue
    margin = (data["revenue"] - data["expense"]) / data["revenue"] * 100
    print(f"{quarter}: {margin:.2f}%")
```

4. Conditional Control Flow (continue) (# 3.3)
if data["revenue"] < 0:: Validates data before attempting mathematical operations (preventing negative or nonsensical financial margins).
continue Statement: Immediately halts execution of the current loop iteration when triggered and jumps straight to the next pass, effectively skipping invalid entries like "Q4".

### Step-by-Step Code Breakdown: Dictionaries & Conditional Logic

#### 1. Single Dictionary Lookup (`# 3.1`)
- **Key-Value Pairs (`{"key": value}`)**: Dictionaries store structured data associated with labeled keys.
- **Key Access (`fin["revenue"]`)**: Values are retrieved by referencing their exact key name inside square brackets `[...]`.

#### 2. Processing a List of Dictionaries (`# 3.2`)
- **Collection of Records**: Storing dictionaries inside a list (`[{}, {}, {}]`) represents tabular data or database records.
- **In-Place Dictionary Mutation**: `fin['margin'] = ...` dynamically creates a new key (`'margin'`) and attaches the computed value directly to each dictionary as the loop runs.

#### 3. Nested Dictionaries & `.items()` (`# 3.3`)
- **Nested Structure**: A dictionary can contain other dictionaries as values (e.g., mapping `"Q1"` to `{"revenue": 50, "expense": 10}`).
- **`.items()` Method**: Returns key-value pairs simultaneously. In `for quarter, data in financials_dict.items():`, `quarter` receives the key string (e.g., `"Q1"`), and `data` receives the inner dictionary.

#### 4. Conditional Control Flow (`continue`) (`# 3.3`)
- **`if data["revenue"] < 0:`**: Validates data before attempting mathematical operations (preventing negative or nonsensical financial margins).
- **`continue` Statement**: Immediately halts execution of the current loop iteration when triggered and jumps straight to the next pass, effectively skipping invalid entries like `"Q4"`.

```python
# 3.1 Single dictionary lookup
fin = {"revenue": 50, "expenses": 10}
profit = fin["revenue"] - fin["expenses"]
margin = profit * 100 / fin["revenue"]
print(f"Single Dict Margin: {margin}%")

# 3.2 List of dictionaries
financials_list = [
    {"revenue": 50, "expenses": 10},
    {"revenue": 60, "expenses": 30},
    {"revenue": 55, "expenses": 35}
]

for fin in financials_list:
    profit = fin["revenue"] - fin["expenses"]
    fin["margin"] = profit * 100 / fin["revenue"]

print("\nUpdated List of Dicts:")
print(financials_list)

# 3.3 Nested dictionary with conditional check
financials_dict = {
    "Q1": {"revenue": 50, "expense": 10},
    "Q2": {"revenue": 60, "expense": 30},
    "Q3": {"revenue": 55, "expense": 35},
    "Q4": {"revenue": -10, "expense": 35},
}

print("\n--- Quarterly Financial Analysis ---")
for quarter, data in financials_dict.items():
    if data["revenue"] < 0:
        print(f"Invalid revenue for {quarter}. Skipping.")
        continue
    margin = (data["revenue"] - data["expense"]) / data["revenue"] * 100
    print(f"{quarter}: {margin:.2f}%")
```
